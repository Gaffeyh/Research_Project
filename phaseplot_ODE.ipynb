```python
##import dependencies
import numpy as np
import matplotlib.pyplot as plt
```

```python
c = -100

def dudt(t, u):
    return c*u +np.sin(t)
```

```python
t0 = 0
tfinal = 10
dt = 0.1

```

```python
t = np.arange(t0,tfinal+dt,dt)
x_range = np.linspace(-0.1,0.1,101)
tm, xm = np.meshgrid(t,x_range)
dxdt_m = dudt(tm, xm)
dt_m = np.ones(tm.shape)

plt.figure()
plt.title('ODE Streamplot')
plt.streamplot(t,x_range,dt_m, dxdt_m)
plt.xlabel('Time')
plt.ylabel('u(t)')
plt.savefig("Graphs_transfer/ODE_Phase.pdf")
```

```python
def AC_ODE(u):
    return u*(1-u**2)
```

```python
t = np.arange(t0,tfinal+dt,dt)
x_range = np.linspace(-2,2,101)
tm, xm = np.meshgrid(t,x_range)
dxdt_m = AC_ODE(xm)
dt_m = np.ones(tm.shape)

plt.figure()
plt.title('Stream Plot AC Forcing Term')
plt.streamplot(t,x_range,dt_m, dxdt_m)
plt.xlabel('Time')
plt.ylabel('u(t)')
plt.savefig("Graphs_transfer/AC_Forcing_Term.pdf")
```

```python
x_range = np.linspace(-1.5,1.5,101)
plt.figure()
plt.plot(x_range, AC_ODE(x_range), label="Forcing term")

# x- and y-axis lines
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)

# grid
plt.grid(True, linestyle="--", alpha=0.6)

plt.annotate("", xy=(-1, 0), xytext=(-1.35, 0),
             arrowprops=dict(arrowstyle="->", color="tab:blue", lw=2))
plt.annotate("", xy=(-1, 0), xytext=(-0.65, 0),
             arrowprops=dict(arrowstyle="->", color="tab:blue", lw=2))
plt.annotate("", xy=(1, 0), xytext=(0.65, 0),
             arrowprops=dict(arrowstyle="->", color="tab:blue", lw=2))
plt.annotate("", xy=(1, 0), xytext=(1.35, 0),
             arrowprops=dict(arrowstyle="->", color="tab:blue", lw=2))


##labelling
plt.title("AC Forcing Term")
plt.xlabel("u")
plt.ylabel("du/dt")
plt.legend()
```
